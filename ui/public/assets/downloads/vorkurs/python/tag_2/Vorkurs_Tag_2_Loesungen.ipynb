## Aufgabe 2.5

Ein Schaltjahr ist ein Jahr, dessen Jahreszahl restlos durch vier teilbar ist. Eine Ausnahme bilden die vollen Jahrhundertjahre, deren Jahreszahl durch 100 teilbar ist: Diese sind nur dann ein Schaltjahr, wenn ihre Jahreszahl auch durch 400 teilbar ist. 

Gib einen booleschen Ausdruck an, der für einen Integer-Wert `jahr` genau dann wahr wird, wenn es sich bei dem in `jahr` gespeicherten Wert um die Jahreszahl eines Schaltjahres handelt.

```python
jahr = 1900

if jahr % 100 == 0:
    if jahr % 400 == 0:
        print("Schaltjahr")
    else:
        print("kein Schaltjahr")
elif jahr % 4 == 0:
    print("Schaltjahr")
else:
    print("Kein Schaltjahr")

# oder in einer Zeile:
if ((jahr % 4 == 0) and (jahr % 100 != 0)) or (jahr % 400 == 0):
    print("Schaltjahr")
```
