### Aufgabe 3.3

Schwierigkeit: 😈😈

Lege ein Programm an, in dem es zunächst folgende Integer Variablen gibt: \
einser, zweier, dreier, vierer, fuenfer \
Diese fünf Variablen beinhalten jeweils die Anzahl der Studenten, die die
entsprechende Note bei einer Klausur erreicht haben. Berechne nun aus diesen
Noten den Notendurchschnitt der Klausur.


````python
einser = 4
zweier = 3
dreier = 8
vierer = 4
fuenfer = 1
anzahl_studenten = 20
notendurchschnitt = (einser*1 + zweier*2 + dreier*3 + vierer*4 + fuenfer*5) / anzahl_studenten
print(notendurchschnitt)
````
