## 2. Datentypen 🔢

Es gibt unterschiedliche Arten von Variablen. In Python muss der Datentyp nicht explizit angegeben werden, da dieser anhand des Wertes, der in die Variable gespeichert wird, automatisch zugewiesen wird

|**Datentyp**|**Beispiel**|
|:--- | :--- | 
|integer (Ganzzahl)|5, 0, -45|
|float (Kommazahl) |2.5,  3.14,  -2.05|
|bool (Wahrheitswert)| True, False|
|string (Zeichenkette) | "Name", 'Eins', "Blau"|

## 4. Konsolen Ein-/Ausgaben 🖨️

Um mit dem Anwender eines Programms zu kommunizieren, können Ein- und Ausgaben auf das Konsolenfenster gemacht werden
Ausgaben auf die Konsole: print()
Eingaben von der Konsole: input()

```python
x = 5
print("Wert von x:", x)
```

In Jupiter Notebook ist diese Funktion überflüssig, da jede Zelle eine eigene Ausgabe hat. 

Optional kann bei einer Eingabe noch eine Nachricht dem Anwender mitgegeben werden. 

```python
eingabe = input("Gebe eine Zahl an:")
```

```python
print("Deine Zahl lautet:", eingabe)
```

## 5. Kommentare ✍️

Kommentare sind kurze Anmerkungen, die den Programmcode erklären. \
Sie beginnen mit '#' und alles was in der Zeile dahinter steht wird vom Python Interpreter nicht interpretiert

```python
# Dies ist ein Kommentar
# Kommentare beginnen mit einem '#'

i = 5   # Variable i wird mit dem Wert 5 initialisiert
```

Kommentare werden vom Programmierer auch häufig dafür verwedet bestimmte Zeilen kurzfristig auszublenden

```python
# i = 3
i
```
