# 1. Schleifen
Schleifen (engl. loop) dienen dazu, einen bestimmten Programmteil mehrfach auszuführen. Dabei kann es sich um
einzelne Programmzeilen aber auch um komplexere Einheiten handeln.

Man unterscheidet oft zwei Schleifen-Arten: `for` und `while`-Schleifen.

### While-Schleifen

Eine while-Schleife wird ausgeführt, solange ihre Bedingung wahr (`True`) ist.
Definiert wird eine solche Schleife durch das Schlüsselwort `while`. 

Syntax:
```python
while bedingung:
    anweisung_1
    anweisung_2
    ...
```
Die Schleifenbedingung muss zu einem booleschen Wert auswerten (als `True` oder `False` sein), ansonsten könnt ihr
hier boolesche Operatoren kombinieren wie ihr möchtet.

Sie wird zu Beginn jedes Durchlaufs geprüft. Solange sie `True` ist, werden die Anweisungen ausgeführt.
Das birgt aber auch ein Fehler, der häufiger auftritt, als einem lieb ist: unendliche Schleifen.

Ein kleines Beispiel:

````python
a = 0
while a >= 0:
    print(a)
    a = a + 1
````

Was fällt auf? 
-> Die Schleife endet nie! 

Wir haben hier eine sogenannte **Endlos-Schleife** erstellt. 
Man muss darauf achten, dass die Schleifenbedingung im Programmverlauf erfüllt wird und die Schleife abbricht. Sonst passiert genau das hier. 

Das gleiche Beispiel mit einer korrekten Bedingung: 

````python
a = 0
while a < 5:
    print (a)
    a = a + 1
````

Die Variable `a` wird auch **Schleifenvariable** oder **Laufvariable** genannt, weil die Ausführung der Schleife von ihrem Wert abhängt. 
Schleifenvariablen bringen uns schon zum nächsten Thema. 

### For-Schleifen

Bei `for`-Schleifen ist im Vorfeld fest, wie oft sie ausgeführt werden. Das geschieht durch eine **Schleifenvariable** (engl. loop counter). 

Syntax: 
```python
for index in range(n):
  anweisung_1
  anweisung_2
  ...
``` 
`index` bezeichnet hier die Laufvariable, und `n` ist die Anzahl Durchläufe diese Schleife machen soll bevor sie abbricht. 

`range(n)` ist eine Python-Funktion, die dazu dient eine Reihe von Zahlen zu generieren. Die Reihe startet bei 0 und geht bis n: `0`,`1`, ... `n-1`. 

Das Beispiel unten ist äquivalent zu dem Beispiel mit `while` weiter oben:

````python
for a in range(5): 
    print (a)
````

`range()` erlaubt es auch, sowohl Start- als auch Endpunkt und die Schrittweise festzulegen: 
* `range(start, stop)`:  Für eine Zahl `z` ist `start` <= `z` < `stop`

````python
for index in range(3, 7):
  print(index)
````

* `range(start, stop, step)`: `step` gibt die Schrittweite an

````python
for index in range(10, -10, -1):
  print(index)
````

Dabei belassen wir es auch erstmal bei Schleifen. Wir kommen später noch eine besondere Art For-Schleifen zurück. Aber erst betrachten wir einige wichtige Datenstrukturen. 

# 2. Listen
Eine Liste dient dazu, mehrere Elemente (z.B. Variablen) gemeinsam zu speichern. 
Listen sind geordnet, sprich die Reihenfolge der Elemente ist fest innerhalb der Liste. 

### Liste erstellen 
Eine Liste kann sehr einfach mit eckigen Klammern `[]` erstellt werden. Als Konvention werden oft die Namen der Listen im Plural geschrieben, da sie ja mehrere Elemente enthalten.

````python
leer = []
print(leer)
````

````python
namen = ['Fred', 'Yasmin', 'Kilian', 'Erkan']
namen
````

Im Gegensatz zu anderen Programmiersprachen erlaubt Python es, mehrere Elemente **unterschiedlichen Typs** in eine Liste zu speichern. 

````python
gemischt = ['Zero', 42, 0.3, False]
gemischt
````

Eine Liste kann auch andere Listen und Objekte enthalten. 

````python
koordinaten = [[49.01594, 8.390639], [48.84941, 2.31333], [30.040034, 31.23529]]
koordinaten
````

### Zugreifen auf Listenelemente

Um auf ein Element in einer Liste zuzugreifen, nutzt man den `[]`-Operator. 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']

apfel = fruechte[0] # Erstes Element
banane = fruechte[1] # Zweites Element
traube = fruechte[3] # Viertes Element

print(apfel)
print(banane)
print(traube)
````

Der Listenindex fängt bei 0 an, weshalb man beim Zugriff mit `list[i]` das `i-1` Element zurückbekommt. 

### Listenelemente hinzufügen, verändern und löschen

Listen sind **veränderbar**, es können Element hinzugefügt, geändert oder auch gelöscht werden, ohne eine neue Liste erstellen zu müssen. 

#### Listenelemente verändern

Listenelemente zu verändern ist möglich durch eine Zuweisung: 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']

fruechte[3] = 'Brombeere' # ersetze das Element an Index 3 (Traube) durch Brombeere
fruechte
````

#### Listenelement hinzufügen

* `.append()` fügt ein Element am Ende ein.

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']
fruechte.append('Birne') 
fruechte
````

*  `.insert()` fügt ein Element an einer bestimmten Stelle ein. Hier muss man einen Index angeben, an welchem eingefügt werden soll. 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']
fruechte.insert(3, 'Kiwi') # füge Kiwi an 4. Stelle ein
fruechte
````

#### Listenelemente entfernen
* `.pop()` entfernt das letzte Element einer Liste und gibt es zurück. 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']

letztes = fruechte.pop()
print(letztes)

fruechte
````

`.pop(n)` entfernt das Element an Index `n`: 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']

geloescht = fruechte.pop(1)
print(geloescht)

fruechte
````

* `.remove()` entfernt einen bestimmten Wert aus der Liste. Wenn er mehrmals vorkommt, dann wird nur das erste Vorkommen entfernt. 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube', 'Maracuja']

fruechte.remove('Maracuja')
fruechte
````

Wenn man versucht, einen Wert zu entfernen, der nicht in der Liste ist, wird ein Fehler ausgegeben. 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']

fruechte.remove('Erdbeere')
fruechte
````

### Suche nach Listenelementen

Mit der `.index()`-Funktion kann man gezielt nach einem Wert in einer Liste suchen. Die Funktion gibt dann den Index des gesuchten Werts in der Liste zurück.

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']
fruechte.index('Traube')
````

# 3. Tupel

Tupel kennt ihr vielleicht aus der Mathematik. 

In Python sind Tupel (engl. tuple) ähnlich zu Listen, sie speichern eine **geordnete** Menge an Elementen. Allerdings können _sie nicht mehr verändert werden_, nachdem sie erstellt sind. Sie sind **unveränderbar**.

Tupel werden mit runden Klammern `()` erstellt: 

````python
staedte = ('Karlsruhe', 'Mannheim', 'Freiburg')
staedte
````

Der Zugriff auf die Elemente passiert analog zu Listen mit den eckigen Klammern `[]`: 

````python
stadt = staedte[1]
stadt
````

Was hingegen nicht geht, ist das Verändern des Tupel und der darin gespeicherten Werte: 

````python
staedte[1] = 'Stuttgart'
````

````python
staedte.append()
````

# 4. Dictionaries

Dictionaries speichern Schlüssel/Wert-Paare (engl. key-value-pairs). Jedem Schlüssel ist mit einem Wert assoziiert. Der Schlüssel darf deshalb **nicht verändert** werden und muss **einzigartig** sein. 
Der Wert kann hingegen beliebigen Typ haben und verändert werden. 

### Dictionary erstellen
Dictionaries werden mit geschweiften Klammern `{}` definiert. Einzelne Paare müssen mit Kommas getrennt werden. 


````python
leeres_dict = {}
leeres_dict
````

````python
studi = {
  'vorname': 'William',
  'nachname': 'Schmidt',
  'matrikelnummer': 80000,
  'zurueckgemeldet': True,
  'bestandene_faecher': ['Informatik 1', 'Mathematik 1']
}
studi
````

````python
studi = {
  'vorname': 'William',
  'vorname': 'Schmidt',
  'matrikelnummer': 80000,
  'zurueckgemeldet': True,
  'bestandene_faecher': ['Informatik 1', 'Mathematik 1']
}
studi
````

### Zugreifen auf Werte
Der Zugriff auf gespeicherte Werte funktioniert über den Schlüssel und mit eckigen Klammern `[]`: 

````python
studi['vorname']
````

### Dictionary verändern
Wie Listen sind Dictionaries **veränderbar**.
#### Ändern eines Wertes
Werte in Dictionaries können ohne Probleme geändert werden. Das geht analog zur Liste mit dem `[]`-Operator und einer Zuweisung. 

Syntax: 
```python
dict[key] = new_value
```

````python
studi = {
  'vorname': 'William',
  'nachname': 'Schmidt',
  'matrikelnummer': 80000,
  'zurueckgemeldet': True,
  'bestandene_faecher': ['Informatik 1', 'Mathematik 1']
}

studi['nachname'] = 'Maier'
studi
````

#### Löschen von Schlüssel-Wert-Paaren

Um einen bestimmten Schlüssel und dessen assoziierten Wert zu löschen wird das `del`-Schlüsselwort verwendet. 

Syntax: 
```python
del dict[key]
```

````python
studi = {
  'vorname': 'William',
  'nachname': 'Schmidt',
  'matrikelnummer': 80000,
  'zurueckgemeldet': True,
  'bestandene_faecher': ['Informatik 1', 'Mathematik 1']
}

del studi['zurueckgemeldet'] 
studi
````

#### Hinzufügen von Schlüssel-Wert-Paaren
Das Hinzufügen von neuen Paaren funktioniert relativ einfach mit einer Zuweisung auf einen neuen Schlüssel. Dieser darf allerdings noch nicht im Dictionary existieren, da sonst bei dem exisitierenden Schlüssel lediglich der Wert überschrieben wird. 

Syntax: 
```python
  dict[new_key] = new_value
```

````python
studi = {
  'vorname': 'William',
  'nachname': 'Schmidt',
  'matrikelnummer': 80000,
  'zurueckgemeldet': True,
  'bestandene_faecher': ['Informatik 1', 'Mathematik 1']
}

studi['geburtstag'] = '2000-10-21'

studi
````

# 5. For-Each Schleifen

For-Each-Schleifen sind eine besondere Art von For-Schleifen, die sich dadurch auszeichnen, dass man damit über Sammlungen von Elementen iterieren kann. Das können zum Beispiel Listen, Tupel oder auch Dictionaries sein. 

Die Syntax ist sehr ähnlich zu normalen For-Schleifen: 
```python
for item in list:
    anweisung_1
    anweisung_2
```
Für jeden Durchlauf wird `item` sukzessive die verschiedenen Elemente der Liste zugewiesen. Innerhalb der Schleife kann man dann jedes Element bearbeiten.

Beispiel: 

````python
fruechte = ['Apfel', 'Banane', 'Maracuja', 'Traube']

for frucht in fruechte:
    print(frucht)
  
````

````python
fruechte
````

Manchmal will man aber trotzdem Indizes in der Schleife haben. Dafür kann man die `enumerate()`-Funktion verwenden. Sie erstellt ein Tupel aus Index und Wert. 

````python
staedte = ('Karlsruhe', 'Mannheim', 'Freiburg')

for stadt in enumerate(staedte): 
  print(stadt)
````

Für Dictionaries muss man angeben, ob man über die Elemente, die Schlüssel oder die Werte iterieren möchte. 

Um über alle Elemente zu iterieren muss man die `.items()`-Funktion verwenden. Diese gibt die key-value-Paare als Tupel zurück. 

````python
studi = {
  'vorname': 'William',
  'nachname': 'Schmidt',
  'matrikelnummer': 80000,
  'zurueckgemeldet': True,
  'bestandene_faecher': ['Informatik 1', 'Mathematik 1']
}

for item in studi.items():
  print(item)
````

Über die Schlüssel zu iterieren funktioniert analog mit der `.keys()`-Funktion. Das ist das Standard-Verhalten wenn man über Dictionaries iteriert, ohne wie hier ein Verhalten zu bestimmen.

````python
for key in studi.keys():
  print(key)
````

Über die Werte zu iterieren geht mit der `.values()`-Funktion: 

````python
for value in studi.values():
  print(value)
````
