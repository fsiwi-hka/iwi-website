#### Aufgabe 3.4.4. 
Schwierigkeit: 😈😈

Alice und Bob möchten jetzt ihren Durchschnitt vergleichen. Errechnet für beide den Durchschnitt und gebt den besseren Durchschnitt aus. Nutzt dafür Schleifen.

Eine Liste mit Alice Noten findet ihr unten:

````python
# Lösung mit einer Schleife, da beide listen die gleiche Länge haben.
alice_noten = [1.0, 1.3, 2.7, 2.0]

alice_durchschnitt = 0
bob_durchschnitt = 0

for index in range(len(alice_noten)): 
    alice_durchschnitt += alice_noten[index]
  
    
for index in range(len(bob_noten)):
    bob_durchschnitt = bob_durchschnitt + bob_noten[index]

alice_durchschnitt = alice_durchschnitt / len(alice_noten)
bob_durchschnitt = bob_durchschnitt / len(bob_noten)

print(alice_durchschnitt)
print(bob_durchschnitt)

if alice_durchschnitt > bob_durchschnitt:
    print(bob_durchschnitt)
elif alice_durchschnitt < bob_durchschnitt: 
    print(alice_durchschnitt)
else: 
    print("Bob und Alice haben den gleichen Durchschnitt.")
````

Die Zahlenfolge 1, 1, 2, 3, 5, 8, 13, . . . nennt man Fibonacci-Zahlen. Dabei entspricht jede Zahl in der Folge gerade der Summe ihrer beiden Vorgänger: 1+1 = 2, 1 + 2 = 3, 2 + 3 = 5, 3 + 5 = 8, . . . Schreibe eine Schleife, die alle Fibonacci-Zahlen ausgibt, die kleiner als 1000 sind.

````python
# Lösung 1 (schneller)
zahl = 1
vorgaenger = 1

while vorgaenger < 1000:
    print(vorgaenger, end='  ')
    vorgaenger, zahl = zahl, zahl + vorgaenger
````

````python
# Lösung 2
z1 = 1
z2 = 1
tempZ = 0
print("Die Fibonacci-Zahlen:\n1")
while z1 + z2 < 1000:
  print(z2+z1)
  tempZ = z1
  z1 = z2
  z2 = z1 + tempZ
````

Nun möchte man erkennen ob ein Wort ein __Palindrom__ ist. Ein Palindrom ist ein Wort das von vorne oder hinten gelesen gleich ist. 

Zum Beispiel: Bob, Anna, Lagerregal, Kayak

Schreibe ein Programm, welches Palindrome erkennt. 

Tipp: mit `.lower()` macht alle Buchstaben in einem String zu kleinen Buchstaben.

````python
str = "Bobob"
str = str.lower()

is_palindrom = True
for i in range(0, int(len(str) / 2)): 
  j = len(str) - i - 1
  if str[i] != str[j]:
    is_palindrom = False

print(is_palindrom)
````

Seien A und B zwei Listen unterschiedlicher Länge: 

```python 

  A = [4, 2, 1, 9, 10]
  B = [5, 3, 1, 0, -1, 7]

```

Du sollst eine Lösung implementieren, die eine Listen, unabhängig von ihrer Länge, aufsteigend sortiert. 
Zum Sortieren vergleicht man zwei nebeneinanderliegende Listen-Elemente und überprüft ob das linke größer als das rechts ist. Ist das der Fall, dann werden die Elemente getauscht. Ansonsten nicht. 

Du musst nur eine Liste auf einmal sortieren. 

Tipp: du brauchst zwei Schleifen. 

````python
A = [4, 2, 1, 9, 10]
B = [5, 3, 1, 0, -1, 7]

len_arr = len(B)
arr = B

# iteriere über alle Listenelemente
for i in range(0, len_arr -1): 
  # iteriere über alle Listenelemente bis zum Vorletzten 
  for j in range(0, len_arr - i - 1): 
    # wenn das linke Element größer als das rechter ist, dann tausche
    if arr[j] > arr[j + 1]:
      arr[j], arr[j + 1] = arr[j + 1], arr[j]
    
      # oder mit temporärer Variable: 
      # t = arr[j]
      # arr[j] = arr[j+1]
      # arr[j+1] = t


print(arr)
````
