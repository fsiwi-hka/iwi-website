## Aufgabe 4.3

Schreibe eine Funktion die drei Zahlen (integers) als als Parmeter übergeben bekommt und die kleinste Zahl zurück gibt.

Der Aufruf *get_min(7,4,8)* soll die Zahl 4 zurückgeben.

```python
def get_min(a, b, c):
    """Gibt das Minimum von drei Zahlen aus.
    :param a: erste Zahl
    :param b: zweite Zahl
    :param c: dritte Zahl
    :return: kleinste der drei Zahlen
    """
    kleinste_zahl = a
    if b < a:
        kleinste_zahl = b
    if c < kleinste_zahl:
        kleinste_zahl = c
    return kleinste_zahl

print(get_min(7, 4, 8))
print(get_min(4, 8, 2))
print(get_min(4, 6, 7))
```

Eine (schönere) Alternative:

```python
def get_min(a, b, c):
    """Gibt das Minimum von drei Zahlen aus.
    :param a: erste Zahl
    :param b: zweite Zahl
    :param c: dritte Zahl
    :return: kleinste der drei Zahlen
    """
    kleinste_zahl = a
    for i in [b, c]:
        if i < kleinste_zahl:
            kleinste_zahl = i
    return kleinste_zahl      

print(get_min(5, 4, 3))
print(get_min(4, 8, 2))
print(get_min(4, 6, 7))
```

Die Funktion *get_rand_num* bekommt eine Zahl als Parameter und soll eine zufälligen Exponenten generieren und diese dann ausgerechnet zurückgeben. Achtung, wenn die Funktion keinen Integer bekommt, dann soll einen String “kein Integer wert” zurückgegeben werden. Verwende diesen Startcode:

```python
import random

def get_rand_num(num):
    """Generiert einen zufälligen Exponenten und 
    berechnet den Potenzwert durch die gegebene Basis.
    :param num: Basis
    :return: Potenzwert
    """
    if isinstance(num, int):
        rand_num = random.randint(1, 100)
        return num ** rand_num
    
    return 'Kein Integer Wert!'

print(get_rand_num(2))
print(get_rand_num(2.))
print(get_rand_num('Hallo'))
```

Eine andere Alternative:

```python
import random

def get_rand_num(num):
    """Generiert einen zufälligen Exponenten und 
    berechnet den Potenzwert durch die gegebene Basis.
    :param num: Basis
    :return: Potenzwert
    """
    if not isinstance(num, int):
        return 'Kein Integer Wert!'
    rand_number = random.randint(1, 100)
    
    return pow(num, rand_number)

print(get_rand_num(2))
print(get_rand_num(2.))
print(get_rand_num('Hallo'))
```

### 4.Z1

Schreibe eine Funktion *mult*, die beliebig viele Eingabewerte annimmt und diese miteinander multipliziert zurückgibt. Wenn kein Wert gegeben wird, soll 0 zurückgegeben werden.

```python
def mult(*args):
    """Multipliziert alle eingegeben Werte und gibt das Ergebnis zurück.
    Bei keiner Eingabe, wird 0 zurückgegeben.
    :param *args: zu multiplizierende Werte
    :return :Ergebnis der Multiplikation
    """
    
    # Gib 0 zurück, falls die Länge der Eingabe 0 beträgt
    if len(args) == 0:
        return 0
    
    # Iteriere durch die eingegeben Werte und mutlipliziere mit dem Zwischenergebnis
    ergebnis = args[0]
    for arg in args[1:]:
            ergebnis = ergebnis * arg
    return ergebnis

print(mult())
print(mult(3))
print(mult(2, 3, 8, 100))
```

### 4.Z2

Erstelle mit NumPy ein Array aus drei Einsen. Erstelle zusätzlich ein lineares Array von 2 bis 10 mit fünf Werten. Konkateniere beide Arrays zum Array *a*. Finde heraus, wie du mit NumPy alle Werte von *a* miteinander multiplizieren kannst. Rufe die Funktion von *4.Z1* außerdem mit den Werten von *a* auf und betrachte das Ergebnis.

```python
import numpy as np

# Arrays erstellen
ones_arr = np.ones(3)
linear_arr = np.linspace(2, 10, 5)
# konkatenieren
a = np.concatenate((ones_arr, linear_arr))

print(a)

# Produkt bilden
print(np.prod(a))
print(mult(*a))
```
