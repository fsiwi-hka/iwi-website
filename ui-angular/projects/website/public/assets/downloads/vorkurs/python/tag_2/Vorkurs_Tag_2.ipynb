# 1. Vergleichsoperatoren
Oft möchte man beim Programmieren zwei Werte miteinander vergleichen. Hierbei helfen _Vergleichsoperatoren_. Viele davon kommen aus der Mathematik, und sollten euch deswegen schon bekannt vorkommen. 

Python kennt sechs Operatoren, die Vergleiche ermöglichen: 
* `<`/`>`: Kleiner/ größer
* `<=`/`>=`: Kleiner gleich/ größer gleich
* `==`: Gleich
* `!=`: Ungleich

Diese Operatoren vergleichen zwei Werte (sowohl Zahlen als auch Strings) und geben ein boolean (`True` oder `False`) zurück. 

Beispiele: 

````python
25 < 50
````

````python
"Hello" == "Hey"
````

````python
32 != 12
````

````python
# Da A vor B im Alphabet kommt, ist A kleiner als B 
"A" < "B"
````

Mit Variablen sieht das dann so aus: 

````python
x = 5
y = 10

x > y 
````

# 2. Logische Operatoren

Um komplexere Abfragen zu konstruieren, braucht man ein Werkzeug um Bedingungen zu verknüpfen. Hier kommen logische Operatoren ins Spiel. 
Python kennt drei verschiedene logische Operatoren: 
  * `and`/`&`:  `und` prüft ob beide Bedingungen gleichzeitig `True` sind. Wenn ja, dann gibt es `True` zurück, sonst `False`.

Folgende Tabelle hilft dabei, zu wissen welches das Ergebnis ist, abhängig von den Werten von `a` und `b`: 

| `a` | `b` | `a and b`| 
|-----|-----|----------| 
| True| True| True| 
| True| False| False| 
| False| True| False| 
| False| False| False| 

````python
Preis = 5.99
Preis > 5 and Preis < 10
````

* `or`/`|`: `or` prüft ob eine der beiden Bedingungen  `True` ist. Wenn ja, dann gibt es `True` zurück, sonst `False`.

| `a` | `b` | `a or b`| 
|-----|-----|----------| 
| True| True| True| 
| True| False| True| 
| False| True| True| 
| False| False| False| 

````python
Preis = 5.99
Preis > 10 or Preis < 20
````

* `not`: `not` invertiert den booleschen Wert einer Bedingung

| `a` |`not a`| 
|-----|-------| 
|True | False | 
|False| True  | 

````python
Preis = 5.99
not Preis < 5
````

### Präzedenz von logischen Operatoren: 

Man kann auch mehrere logische Operatoren hintereinander verketten. Die Operatoren haben aber eine unterschiedlich hohe Bindung (oder Präzedenz): 
Die höchste Präzedenz hat `not`, danach kommt `and` und zuletzt `or`. 

Python wertet die Operatoren mit der höchsten Präzedenz zuerst aus und geht dann weiter. 

Sprich, folgender Ausdruck wird so ausgewertet: 
```python
not a and b or c 
```
---> 
```python
(not a) and b or c 
```
--> 
```python
((not a) and b) or c 
```

````python
Preis = 5.99
not (Preis < 5 and Preis > 10)
````

````python
not Preis < 5 and Preis > 10
````

In diesem Sinne: 

❗ **Klammern sind wichtig** ❗

# 3. If-Abfrage

### Klassische If-Abfrage 
Die `if`-Abfrage erlaubt es, einen Code-Block nur auszuführen, wenn eine bestimmte Bedingung erfüllt ist.

Die Syntax ist: 
```python
  if Bedingung: 
    anweisung_1()
    ...
```
Achtet immer auf die Einrückung! Sonst verhält sich der Code eventuell nicht so wie ihr möchtet. 

Beispiel:

````python
age = 19
if age >= 18: 
  print("Du bist volljährig!")
````

### If-Else

`if-else` wird benutzt um eine Fallunterscheidung zu machen, also einen Teil des Codes ausführen wenn die Bedingung wahr ist, und ein anderer Teil wenn sie falsch ist. 

Die Syntax ist: 
```python
if Bedingung: 
  if_block
else: 
  else_block
```

Wenn die Bedingung `True` ist, dann wird der `if`-Block ausgeführt. Wenn die Bedingung `False` ist, dann wird der `else`-Block ausgeführt.

````python
age = 17
if age >= 18: 
  print("Du bist volljährig!")
else: 
  print("Du bist minderjährig!")
````

### If-Elif-Else
Wenn man mehrere verschiedene Bedingungen abfragen möchte und für jede Bedingung eine bestimmten Code-Block ausführen möchte, kann man das `if...elif...else`-Konstrukt verwenden. 
`elif` steht dabei für `else if`. 

Syntax: 
```python
if Bedingung_1: 
  if_block
elif Bedingung_2: 
  elif_block_1
elif Bedingung_3: 
  elif_block_2
...
else: 
  else_block
```

Die Bedingungen werden in der Reihenfolge in der sie geschrieben sind abgefragt. Sobald eines `True` ist, wird der zugehörige Code-Block ausgeführt und der Rest wird übersprungen. 
Wenn keine Bedingung `True` ist, dann wird der `else`-Block ausgeführt. 

`else` ist optional und kann auch weggelassen werden. 

````python
age = 12
if age >= 18: 
  print("Du bist volljährig!")
elif age >= 16: 
  print("Du darfst Bier kaufen.")
elif age >= 14:
  print("Du darfst Bier trinken.")
else: 
  print("Du darfst kein Alkohol trinken.")
````

### Switch-Case

Das Switch-Case-Konstrukt ist ein Spezialfall von einem If-Elif-Else. 

Es besteht aus einer Bedingung und mehreren Fällen, die überprüft werden, ob sie die Bedingung erfüllen. Jeder Fall entspricht einem bestimmten Wert. 
Das Input wird gegen diese Werte geprüft und der Fall, dem es entspricht, wird ausgeführt.

Beispiel: 

````python
wochentag = "Dienstag"

if wochentag == "Montag": 
  print("Heute ist der erste Tag der Woche.")
elif wochentag == "Dienstag": 
  print("Heute ist der zweite Tag der Woche.")
elif wochentag == "Mittwoch": 
  print("Heute ist der dritte Tag der Woche.")
elif wochentag == "Donnerstag": 
  print("Heute ist der vierte Tag der Woche.")
elif wochentag == "Freitag": 
  print("Heute ist der fünfte Tag der Woche.")
elif wochentag == "Samstag": 
  print("Heute ist der sechste Tag der Woche.")
elif wochentag == "Sonntag": 
  print("Heute ist der siebte Tag der Woche.")
````
